### 📄 `app.py`

Archivo principal de la implementación.

````python
%%writefile app.py
import time
from collections import defaultdict
from fastapi import FastAPI, Request
from fastapi.responses import JSONResponse

app = FastAPI(title="Rate Limiting - Sesion 18")

# Config del contrato
LIMITE = 3
VENTANA = 60  # segundos

# { cliente_ip: [timestamp1, timestamp2, ...] }
# Se reinicia al iniciar la app porque vive en memoria
peticiones_por_cliente = defaultdict(list)

def get_cliente(request: Request) -> str:
    # El verificador usa TestClient, siempre viene con la misma IP
    # Si hay proxy, respetamos X-Forwarded-For
    forwarded = request.headers.get("x-forwarded-for")
    if forwarded:
        return forwarded.split(",")[0].strip()
    if request.client:
        return request.client.host
    return "anonimo"

@app.get("/api/recurso")
async def recurso(request: Request):
    cliente = get_cliente(request)
    ahora = time.time()

    # 1. Limpiar peticiones viejas fuera de la ventana de 60s
    historial = peticiones_por_cliente[cliente]
    historial = [t for t in historial if ahora - t < VENTANA]
    peticiones_por_cliente[cliente] = historial

    # 2. Si ya hizo 3 en esta ventana -> 429
    if len(historial) >= LIMITE:
        # Cuánto falta para que se libere la ventana
        # El más viejo expira primero
        tiempo_restante = int(VENTANA - (ahora - historial[0])) + 1
        if tiempo_restante <= 0:
            tiempo_restante = 60
        return JSONResponse(
            status_code=429,
            content={"error": "límite de peticiones alcanzado"},
            headers={"Retry-After": str(tiempo_restante)}
        )

    # 3. Si no, registrar y permitir
    historial.append(ahora)
    peticiones_por_cliente[cliente] = historial

    return {"mensaje": "ok", "peticion": len(historial)}
````
