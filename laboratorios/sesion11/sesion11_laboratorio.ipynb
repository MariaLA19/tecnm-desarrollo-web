### 📄 `app.py`

Punto de entrada de la aplicación.

````python
%%writefile app.py
from fastapi import FastAPI
from router import router

app = FastAPI(title="Biblioteca por capas")

app.include_router(router)
````

### 📄 `repositorio.py`

Capa de acceso a datos.

````python
%%writefile repositorio.py
# Capa de acceso a datos. No conoce HTTP.

class RepositorioLibros:
    def __init__(self):
        self._libros = []
        self._siguiente_id = 1

    def listar(self):
        return list(self._libros)

    def crear(self, titulo: str, anio: int):
        libro = {
            "id": self._siguiente_id,
            "titulo": titulo,
            "anio": anio
        }
        self._libros.append(libro)
        self._siguiente_id += 1
        return libro

# Instancia única que comparten servicio y router
repositorio = RepositorioLibros()
````

### 📄 `servicio.py`

Capa de lógica de negocio.

````python
%%writefile servicio.py
# Capa de lógica de negocio: crear y listar libros.
from repositorio import repositorio

def listar_libros():
    return repositorio.listar()

def crear_libro(titulo: str, anio: int):
    return repositorio.crear(titulo=titulo, anio=anio)
````

### 📄 `router.py`

Capa de exposición HTTP.

````python
%%writefile router.py
from fastapi import APIRouter, status
from pydantic import BaseModel
from servicio import listar_libros, crear_libro

router = APIRouter()

class LibroCrear(BaseModel):
    titulo: str
    anio: int

@router.get("/api/libros")
def get_libros():
    return {"libros": listar_libros()}

@router.post("/api/libros", status_code=status.HTTP_201_CREATED)
def post_libro(libro: LibroCrear):
    return crear_libro(titulo=libro.titulo, anio=libro.anio)
````
